## Worked: the bridge of CHN-17

Expected at home = the partner's FOB × (1 + the wedge). Accounted for = our CIF + timing + the hub. Left unexplained = expected − accounted for; its share = left ÷ expected. The like-for-like ratio puts both sides on FOB.

```python
c = mirror[mirror.corridor == 'CHN-17'].copy()
c['expected'] = c.partner_exports_fob_usd * (1 + c.cif_fob_adj)
c['accounted'] = c.home_imports_cif_usd + c.timing_usd.fillna(0) + c.hub_attributed_usd.fillna(0)
c['share'] = (c.expected - c.accounted) / c.expected
c['ratio'] = (c.home_imports_fob_usd.fillna(c.home_imports_cif_usd / (1 + c.cif_fob_adj)) + c.timing_usd.fillna(0) + c.hub_attributed_usd.fillna(0)) / c.partner_exports_fob_usd
c[['year', 'expected', 'accounted', 'share', 'ratio']]
```
