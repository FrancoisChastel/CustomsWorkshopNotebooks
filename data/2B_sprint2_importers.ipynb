The declaration lines of months 1 to 18, each with its price as a share of its peers' typical price (computed for you). **CHN-85 is worked.** You build your corridor's importer table: lines, value, share, the five largest, and the two habits.

*Published customs datasets record an irregularity on 1 to 8 lines in 100; in this lab the random draw finds 3 in 100.*

```python
# Run this cell first (click it, then Shift + Enter). It finds the lab data.
import hashlib, os
from pathlib import Path
import numpy as np, pandas as pd

CANDIDATES = [os.environ.get('WORKSHOP_DATA', ''), 'data', '../data', '/workspace/data', '.']
DATA = next(Path(c) for c in CANDIDATES if c and (Path(c) / 'tariff.csv').exists())
print('Data folder:', DATA)

def _h(v, nd):
    return hashlib.sha256(f'{round(float(v), nd):.{nd}f}'.encode()).hexdigest()[:16]
def _hs(v):
    return hashlib.sha256(str(v).strip().lower().encode()).hexdigest()[:16]
def check(name, value, expected, nd=None):
    if callable(value):
        try:
            value = value()
        except Exception:
            value = ...
    try:
        ok = value is not ... and (_hs(value) == expected if nd is None else _h(value, nd) == expected)
    except Exception:
        ok = False
    shown = value if nd is None or value is ... else f'{value:,.{nd}f}'
    print(('✓ ' if ok else '✗ not yet: ') + name + ' = ' + str(shown))
```

```python
lines = pd.read_csv(DATA / 'features_lines.csv', usecols=['line_id', 'period_month', 'importer_id', 'exporter_id', 'origin', 'chapter', 'customs_value_lcu', 'share_of_median'], low_memory=False)
lines['corridor'] = lines.origin + '-' + lines.chapter.astype(str)
helper = pd.read_csv(DATA / 'corridor_importers.csv')   # computed for you: the median share of peers and the fingerprints
h1 = lines[lines.period_month <= 18]
```

## Worked: CHN-85

```python
g = h1[h1.corridor == 'CHN-85'].groupby('importer_id').agg(lines=('line_id', 'size'), value=('customs_value_lcu', 'sum'),
                                                       low=('share_of_median', lambda s: (s < 0.75).mean()))
g['share'] = g.value / g.value.sum()
g = g.sort_values('value', ascending=False)
g = g.join(helper[helper.corridor == 'CHN-85'].set_index('importer_id')[['med_ratio_h1']])
g['habit'] = ((g.lines >= 5) & (g.med_ratio_h1 < 0.8)).astype(int)                  # habitually priced low
g['levels'] = ((g.lines >= 8) & g.low.between(0.25, 0.75)).astype(int)              # two price levels
print('five largest carry', round(g.share.head(5).sum(), 3))
g.head(8)
```
